```python
# Task 23: Duplicate Data Removal

## Objective
Identify duplicate records in a dataset and remove them using
Pandas to improve data quality.

## Deliverables
- Number of duplicate records
- Display of duplicate records
- Cleaned dataset
- Before-and-after row counts
- Explanation of duplicate removal
```

```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
data = {
    "student_id": [
        1001, 1002, 1003, 1004, 1005,
        1006, 1007, 1008, 1002, 1005, 1004
    ],

    "name": [
        "Aarav", "Diya", "Kabir", "Ananya", "Ishaan",
        "Meera", "Rohan", "Sara", "Diya", "Ishaan", "Ananya"
    ],

    "department": [
        "Computer Science", "Electronics", "Mechanical",
        "Computer Science", "Civil", "Electronics",
        "Mechanical", "Computer Science", "Electronics",
        "Civil", "Computer Science"
    ],

    "age": [
        20, 21, 20, 22, 21, 20, 22, 21, 21, 21, 22
    ],

    "marks": [
        86, 91, 74, 95, 68, 88, 79, 93, 91, 68, 95
    ],

    "attendance": [
        92, 88, 79, 96, 72, 90, 85, 94, 88, 72, 96
    ]
}

df = pd.DataFrame(data)

print("Original dataset created successfully!")
print("Total records:", len(df))

display(df)
```

```python
# Save the original dataset
df.to_csv(
    "student_records_with_duplicates.csv",
    index=False
)

# Load the dataset using Pandas
df = pd.read_csv("student_records_with_duplicates.csv")

print("CSV dataset loaded successfully!")
display(df)
```

```python
print("DUPLICATE RECORD IDENTIFICATION")

duplicate_check = df.duplicated()

print(duplicate_check)

print("\nTotal duplicate records:", duplicate_check.sum())
```

```python
print("ALL DUPLICATE-RELATED RECORDS")

duplicate_rows = df[
    df.duplicated(keep=False)
]

display(duplicate_rows)

print(
    "Total rows belonging to duplicate groups:",
    len(duplicate_rows)
)
```

```python
print("DUPLICATES BASED ON STUDENT ID")

duplicate_ids = df[
    df.duplicated(
        subset=["student_id"],
        keep=False
    )
]

display(duplicate_ids)
```

```python
print("REMOVING DUPLICATE RECORDS")

before_cleaning = len(df)

cleaned_df = df.drop_duplicates()

after_cleaning = len(cleaned_df)

removed_records = before_cleaning - after_cleaning

print("Records before cleaning:", before_cleaning)
print("Records after cleaning:", after_cleaning)
print("Duplicate records removed:", removed_records)

display(cleaned_df)
```

```python
print("VERIFYING CLEANED DATASET")

remaining_duplicates = cleaned_df.duplicated().sum()

print("Remaining duplicate records:", remaining_duplicates)
print("Final number of records:", len(cleaned_df))

if remaining_duplicates == 0:
    print("Dataset successfully cleaned!")
else:
    print("Some duplicates still remain.")

display(cleaned_df)
```

```python
comparison = pd.DataFrame({
    "Description": [
        "Records before cleaning",
        "Duplicate records removed",
        "Records after cleaning",
        "Remaining duplicate records"
    ],

    "Count": [
        before_cleaning,
        removed_records,
        after_cleaning,
        remaining_duplicates
    ]
})

display(comparison)
```

```python
# Save duplicate-related records
duplicate_rows.to_csv(
    "duplicate_records.csv",
    index=False
)

# Save the cleaned dataset
cleaned_df.to_csv(
    "cleaned_student_records.csv",
    index=False
)

# Save before-and-after comparison
comparison.to_csv(
    "duplicate_summary.csv",
    index=False
)

print("All project CSV files exported successfully!")
```
